```sql
with construcots_metrics as
(
    select constructorname, 
        sum(race_starts) as race_starts,
        sum(number_of_podiums) as total_podiums,
        sum(number_of_wins) as total_wins ,
        sum(case when standings = 1 then 1 else 0 end) as championships
from formula1.gold.constructors_standing
group by constructorname
having championships >= 1
)
select constructorname,
       race_starts,
       total_wins,
       total_podiums,
       championships,
       (championships * 100) + (total_wins * 10) + (total_podiums * 3) as score
       from construcots_metrics
order by score desc

```
