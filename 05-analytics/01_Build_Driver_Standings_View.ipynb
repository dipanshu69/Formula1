```python
%sql

create or replace view formula1.gold.driver_standing
as
with cte as (
select r.season as season,
        d.driver_id,
        d.driver_name,
        d.nationality,
        count(*) as race_starts,
        sum(r.points) as total_points,
        count_if(r.is_win) as number_of_wins,
        count_if(r.is_podium) as number_of_podiums
from formula1.gold.fact_results r
join formula1.gold.dim_drivers d
on r.driver_id == d.driver_id
group by r.season, d.driver_id, d.driver_name, d.nationality
)
select rank() over (partition by season order by total_points desc, number_of_wins desc) as rank, * 
from cte
```

```python
%sql
create or replace view formula1.gold.constructors_standing
as
with cte as (
select  r.season as season,
        c.constructor_id as constructorid,
        c.constructor_name as constructorname,
        c .nationality as nationality,
        count(*) as race_starts,
        sum(r.points) as total_points,
        count_if(r.is_win) as number_of_wins,
        count_if(r.is_podium) as number_of_podiums
from formula1.gold.fact_results r
join formula1.gold.dim_constructors c
on r.constructor_id == c.constructor_id
group by season, constructorid, constructorname, nationality
)
select 
rank() over (partition by season order by total_points desc, number_of_podiums desc) as standings, * 
from cte
```
